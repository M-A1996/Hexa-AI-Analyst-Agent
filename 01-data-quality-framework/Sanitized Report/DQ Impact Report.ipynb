# Data-Quality Impact Report &mdash; before vs after automated checks

**Project:** Automated Data-Quality framework for the **HEXA VPN** app &nbsp;&middot;&nbsp; **Built at:** Hazel Mobile &nbsp;&middot;&nbsp; **Data shown:** synthetic / illustrative

<div style="background:#eff6ff;border:1px solid #bfdbfe;color:#1e40af;padding:10px 14px;border-radius:8px;font-family:sans-serif;font-size:13px"><b>Data-quality framework I built at Hazel Mobile for the HEXA VPN app.</b> The figures below are <b>synthetic / illustrative</b> for portfolio use &mdash; they show the <i>method</i> (an automated data-quality before/after model), <b>not real client, user, or production data</b>.</div>

**What this shows:** an automated data-quality framework that reclassifies telemetry to reveal the *true* health of a metric. The checks **never delete events** &mdash; a caught event is **reclassified** success &rarr; failure, and users left with no clean success drop out.

**Affected users are split three ways:**
- **One and done (Type A)** &mdash; a single session ever.
- **One and done (Type B)** &mdash; several sessions but active on one calendar day.
- **Retained** &mdash; active on 2+ days (the users worth acting on).

*Self-validation (self-check statement): all cross-checks PASS.*
