## Action 1 — Handshake / probe timeout (40.7% of failures)


• Lower the initial handshake timeout and add a fast retry on an alternate outbound before giving up.  • Race 2–3 protocols in parallel on connect and keep the first to complete.  • Check MTU / MSS clamping on mobile networks (common cause of stalled handshakes).

## Action 2 — Read timeout mid-session (26.4%)


• Enable protocol keep-alive / persistent-keepalive so idle NAT mappings don't drop.  • Shorten server-side idle timeouts and reconnect seamlessly.  • Investigate load on the servers with the highest mid-session drop share.

## Action 3 — TLS / cert reject (15.4%)


• Verify device clock-skew handling (expired-before-valid errors).  • Refresh the CA bundle and confirm cert-pinning matches the current chain.  • Roll certs ahead of expiry with overlap.
